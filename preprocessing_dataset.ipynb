```python
from sklearn.utils import shuffle
```

```python
num_sample = application_train.shape[0]
```

```python
testing_set = application_train.sort_values(["SK_ID_CURR"]).iloc[-int(0.1*num_sample):, :]

_temp = application_train.sort_values(["SK_ID_CURR"]).iloc[:-int(0.1*num_sample), :]
_temp = shuffle(_temp, random_state=0)
validation_set = _temp.iloc[-int(0.1*num_sample):, :]
training_set = _temp.iloc[:-int(0.1*num_sample), :]

del _temp
```

```python
testing_set.shape, validation_set.shape, training_set.shape
```

```python
training_set.to_csv("./dataset/training_set.csv", index=False)
validation_set.to_csv("./dataset/validation_set.csv", index=False)
testing_set.to_csv("./dataset/testing_set.csv", index=False)
```
