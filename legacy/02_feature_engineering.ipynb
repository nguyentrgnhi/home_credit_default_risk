```python
mapping_education_type = {
    'Lower secondary': 0,
    'Secondary / secondary special': 1,
    'Incomplete higher': 2,
    'Higher education': 3,
    'Academic degree': 4
}

def preprocess_education_type(input_str):
    return mapping_education_type[input_str]

def preprocess(dataset):
    df = dataset.copy()
    # gender
    df["CODE_GENDER"] = (df["CODE_GENDER"] == "F").astype(int)
    # car
    df["FLAG_OWN_CAR"] = (df["FLAG_OWN_CAR"]=="Y").astype(int)
    # realty
    df["FLAG_OWN_REALTY"] = (df["FLAG_OWN_REALTY"]=="Y").astype(int)
    # education
    df["NAME_EDUCATION_TYPE"] = df["NAME_EDUCATION_TYPE"].apply(preprocess_education_type)
    # income type
    df["NAME_INCOME_TYPE_COM"] = (df["NAME_INCOME_TYPE"] == "Commercial associate").astype(int)
    df["NAME_INCOME_TYPE_PEN"] = (df["NAME_INCOME_TYPE"] == "Pensioner").astype(int)
    df["NAME_INCOME_TYPE_STA"] = (df["NAME_INCOME_TYPE"] == "State servant").astype(int)
    df["NAME_INCOME_TYPE_WOR"] = (df["NAME_INCOME_TYPE"] == "Working").astype(int)
    # age
    df["DAYS_BIRTH"] = (df["DAYS_BIRTH"] / 365).astype(int).abs()

    return df
```

```python
bureau = pd.read_csv("bureau.csv")
bureau = bureau.groupby(["SK_ID_CURR"], as_index=False).agg({"CREDIT_DAY_OVERDUE": "max"})
```

```python
training_set = pd.read_csv("./dataset/training_set.csv")
validation_set = pd.read_csv("./dataset/validation_set.csv")
testing_set = pd.read_csv("./dataset/testing_set.csv")
```

```python
training_set = preprocess(training_set)
validation_set = preprocess(validation_set)
testing_set = preprocess(testing_set)
```

```python
training_set = pd.merge(training_set, bureau, on=["SK_ID_CURR"], how="left").fillna({"CREDIT_DAY_OVERDUE": 0})
validation_set = pd.merge(validation_set, bureau, on=["SK_ID_CURR"], how="left").fillna({"CREDIT_DAY_OVERDUE": 0})
testing_set = pd.merge(testing_set, bureau, on=["SK_ID_CURR"], how="left").fillna({"CREDIT_DAY_OVERDUE": 0})
```

```python
used_columns = ['SK_ID_CURR', 'TARGET', 'CODE_GENDER', 'FLAG_OWN_CAR', 'FLAG_OWN_REALTY',
                'NAME_EDUCATION_TYPE', 'NAME_INCOME_TYPE_COM', 'NAME_INCOME_TYPE_PEN',
                'NAME_INCOME_TYPE_STA', 'NAME_INCOME_TYPE_WOR', 'DAYS_BIRTH',
                'CNT_CHILDREN', 'AMT_INCOME_TOTAL', 'AMT_CREDIT', 'AMT_ANNUITY','FLAG_EMP_PHONE', 
               'REG_CITY_NOT_LIVE_CITY', 'REGION_RATING_CLIENT_W_CITY'  ]
```

```python
training_set = training_set.fillna({"AMT_ANNUITY": 0})
validation_set = validation_set.fillna({"AMT_ANNUITY": 0})
testing_set = testing_set.fillna({"AMT_ANNUITY": 0})
```

```python
training_set[used_columns].to_csv("./dataset/final/training_set.csv", index=False)
validation_set[used_columns].to_csv("./dataset/final/validation_set.csv", index=False)
testing_set[used_columns].to_csv("./dataset/final/testing_set.csv", index=False)
```
